```python
import pandas as pd

# Load from parquet instead of CSV — parquet preserves the conversation structure exactly.
# The CSV version breaks when conversation text contains apostrophes (e.g. "Here's", "I'd").
sample_tagged = pd.read_parquet("../Data/WildChatData/wildchat_sample_5000_with_conv.parquet")

print(f"Loaded {len(sample_tagged):,} rows")
print(f"Columns: {sample_tagged.columns.tolist()}")
```

```python
TOPIC_KEYWORDS = {
    "coding":           ["python", "javascript", "code", "function", "bug", "error", "programming", "script", "html", "css", "sql"],
    "math":             ["calculate", "equation", "solve", "algebra", "calculus", "integral", "derivative", "probability"],
    "writing":          ["essay", "poem", "story", "write", "paragraph", "draft", "edit", "summarize", "letter"],
    "creative":         ["imagine", "fictional", "character", "plot", "fantasy", "roleplay", "scenario"],
    "question_answer":  ["what is", "how does", "explain", "define", "why is", "tell me about"],
    "work_business":    ["email", "resume", "job", "meeting", "proposal", "report", "presentation"],
    "health":           ["symptom", "medicine", "doctor", "diet", "exercise", "mental health", "anxiety"],
    "grammar":          ["fix grammar", "proofread", "correct the grammar", "convert to third-person", "correct my", "fix my", "grammar check"],
    "gaming":           ["gta", "roblox", "board game", "pokemon", "minecraft", "video game", "playstation", "xbox", "steam", "game hack"],
    "education":        ["mla", "apa", "citation", "seminar", "homework", "assignment", "thesis", "textbook", "lecture", "academic"],
    "image_generation": ["photoshoot", "generate an image", "create a picture", "draw a scene", "stable diffusion", "midjourney", "dall-e", "image prompt"],
    "translation":      ["translate", "translation", "in english", "in spanish", "in french", "in german", "in chinese", "in japanese", "in russian", "in arabic"],
    "technology":       ["how do i download", "how do i install", "how to set up", "software", "hardware", "ai tool", "app", "website", "browser", "server"],
    "casual":           ["hi there", "hello", "hey", "good morning", "good evening", "how are you", "what's up", "nice to meet"],
}
```

```python
def get_tags(messages):
    # Parquet stores conversation as numpy.ndarray, so accept both list and ndarray
    if not isinstance(messages, (list, __import__("numpy").ndarray)):
        return ["untagged"]

    full_text = " ".join(
        msg.get("content", "") if isinstance(msg, dict) else str(msg)
        for msg in messages
    ).lower()

    tags = [topic for topic, keywords in TOPIC_KEYWORDS.items()
            if any(kw in full_text for kw in keywords)]

    return tags if tags else ["untagged"]

sample_tagged["tags"] = sample_tagged["conversation"].apply(get_tags)
display(sample_tagged[["conversation_hash", "model", "tags"]].head(10))
```

```python
# Inspect what the conversation column actually looks like
first = sample_tagged["conversation"].iloc[0]
print("Type:", type(first))
print()
print("Value:", first)
```

```python
def print_conversation(index):
    row = sample_tagged.iloc[index]
    messages = row["conversation"]
    print(f"--- Conversation {index} | Model: {row['model']} | Tags: {row.get('tags', 'not yet tagged')} ---\n")
    for msg in messages:
        role = msg.get("role", "unknown").upper()
        content = msg.get("content", "")
        print(f"[{role}]:\n{content}\n")
    print("=" * 60)

```

```python
for i in range(3):
    print_conversation(35)
```
